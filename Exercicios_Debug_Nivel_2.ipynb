**Exercício 6:**

**Exercício 10:**
O código tenta ordenar um DataFrame pela coluna 'Nota', mas a coluna não existe. Corrija o código para que ele ordena pela coluna 'Pontos' (que é a correta).

*Dica: O erro é `KeyError: 'Nota'`.*

```python
import pandas as pd

dados = {
    "Nome": ["Ana", "Bruno", "Carla"],
    "Pontos": [15, 18, 12]
}

df = pd.DataFrame(dados)

df_ordenado = df.sort_values(by="Pontos")

print(df_ordenado)
```
